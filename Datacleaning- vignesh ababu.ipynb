```python
#Heatmap colors temparture 
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("eda_sample_dataset.csv")

data = df[
    ["Age", "Study_Hours", "Attendance",
     "Marks", "Internet_Usage_Hours"]
]

corr = data.corr()

plt.figure(figsize=(8, 6))

plt.imshow(corr, cmap="coolwarm")


plt.xticks(
    range(len(corr.columns)),
    corr.columns,
)

plt.yticks(
    range(len(corr.columns)),
    corr.columns
)

for i in range(len(corr.columns)):
    for j in range(len(corr.columns)):
        plt.text(
            j, i,
            f"{corr.iloc[i, j]:.2f}",
            ha="center",
            va="center"
        )

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
department_count = df["Department"].value_counts()

department_count
```

```python
department_count.plot(
    kind="bar",
    figsize=(7, 4),
    color="green"
)

plt.title("Number of Students by Department")
plt.xlabel("Department")
plt.ylabel("Number of Students")
plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
gender_department = pd.crosstab(
    df["Department"],
    df["Gender"]
)

gender_department
```

```python
gender_department.plot(
    kind="bar",
    figsize=(8, 5),
    color=["pink", "steelblue"]
    
)

plt.title("Gender Distribution by Department")
plt.xlabel("Department")
plt.ylabel("Number of Students")
plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
average_marks = df.groupby("Department")["Marks"].mean()

average_marks
```

```python
average_marks.plot(
    kind="bar",
    figsize=(7, 4),
    color=('green')
)

plt.title("Average Marks by Department")
plt.xlabel("Department")
plt.ylabel("Average Marks")
plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
avg_marks_gender = df.groupby(
    ["Department", "Gender"]
)["Marks"].mean()

avg_marks_gender
```

```python
avg_marks_gender.unstack().plot(
    kind="bar",
    figsize=(8, 5),
    color=["pink", "steelblue"]
)

plt.title("Average Marks by Department and Gender")
plt.xlabel("Department")
plt.ylabel("Average Marks")
plt.xticks(rotation=0)
plt.legend(title="Gender")
plt.grid(axis="y", alpha=0.1)

plt.show()
```

```python
gender_count = df["Gender"].value_counts()


plt.figure(figsize=(6, 6))

plt.pie(
    gender_count,
    labels=gender_count.index,
    autopct="%1.1f%%",
    startangle=90,
    colors=["steelblue", "pink"]
)

plt.title("Gender Percentage")
plt.show()
```
